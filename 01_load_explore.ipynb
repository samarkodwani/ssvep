```python
import numpy as np, scipy.io as sio
mat = sio.loadmat('data/S1.mat')
data = mat['data']
print(data.shape)
```

```python
freq_phase = sio.loadmat('data/Freq_Phase.mat')
freq_phase['freqs']
```

```python
import mne
X = data.transpose(2,3,0,1).reshape(240, 64, 1500) #why this?
labels = np.repeat(np.arange(40), 6)  #why this also
ch_names = []
with open('data/64-channels.loc') as f:
    for line in f:
        parts = line.split()
        if len(parts) >= 4:
            ch_names.append(parts[3])

print(len(ch_names), ch_names)   # must be 64
info = mne.create_info(ch_names, sfreq = 250, ch_types= 'eeg')
events = np.column_stack([np.arange(240), np.zeros(240, int), labels])
epochs = mne.EpochsArray(X*1e-6, info, events = events, tmin = 0.5)
epochs.drop_channels(['CB1', 'CB2'])
epochs.set_montage(mne.channels.make_standard_montage('standard_1005'),
                   match_case=False)
print(len(epochs.ch_names))   
```

```python
print(epochs)              # should say 240 events, -0.5 to ~5.496 s
print(epochs.get_data().shape)   # (240, 64, 1500)
print(epochs.times[[0, -1]])
```
